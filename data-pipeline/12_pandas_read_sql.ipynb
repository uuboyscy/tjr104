```python
from urllib.parse import quote_plus

import pandas as pd
from sqlalchemy import create_engine

username = quote_plus("root")
password = quote_plus("1234567890")
server = "127.0.0.1:3308"
db_name = "TESTDB"

conn = create_engine(
    f"mysql+pymysql://{username}:{password}@{server}/{db_name}",
).connect()
```

```python
quote_plus("@")
```

```python
df = pd.read_sql(
    "select * from Staff",
    conn,
)
df
```

```python
df["ID"].str.replace("00", "01")
```

```python
df["ID"] = df["ID"].str.replace("00", "01")
```

```python
df
```

```python
df.to_sql(
    "Staff",
    conn,
    index=False,
    if_exists="append",
)
```

```python
conn.commit()
```
