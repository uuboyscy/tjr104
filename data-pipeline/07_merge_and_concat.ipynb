```python
import pandas as pd

staff_detail_different_col_path = "staff_detail_different_col.csv"
staff_detail_path = "staff_detail.csv"
staff_samecol_path = "staff_samecol.csv"
staff_path = "staff.csv"

staff_detail_different_col_df = pd.read_csv(staff_detail_different_col_path)
staff_detail_df = pd.read_csv(staff_detail_path)
staff_samecol_df = pd.read_csv(staff_samecol_path)
staff_df = pd.read_csv(staff_path)
```

```python
staff_detail_df.merge(staff_df)
```

```python
staff_detail_different_col_df.merge(
    staff_df,
    how="left",
    left_on="EmpNum",
    right_on="ID"
)
```

```python
staff_detail_different_col_df.merge(
    staff_samecol_df,
    how="left",
    left_on="EmpNum",
    right_on="ID",
    suffixes=["_fact", "_dim"]
)
```

```python
import pandas as pd

df = pd.DataFrame(
    data=[
        ["Allen", 175, 22],
        ["Ted", 180, 25]
    ],
    columns=["Name", "Height", "Age"]
)

df_1 = pd.DataFrame(
    data=[
        ["Jake", 177, 28],
        ["Mary", 170, 24]
    ],
    columns=["Name", "Height", "Age"]
)
```

```python
"""
select * from df
union
select * from df_1
"""
pd.concat(
    [df, df_1],
)
```

```python
pd.concat(
    [df, df_1],
).reset_index()
```

```python
pd.concat(
    [df, df_1],
).reset_index(drop=True)
```

```python
pd.concat(
    [df, df_1],
    ignore_index=True,
)
```
